```python
# We will start with one version of parameters
mcorr_params =\
{
  'main': # this key is necessary for specifying that these are the "main" params for the algorithm
    {
        'max_shifts': [8, 8],
        'strides': [64, 64],
        'overlaps': [32, 32],
        'max_deviation_rigid': 3,
        'border_nan': 'copy',
        'pw_rigid': True,
        'gSig_filt': None
    },
}
```

```python
df.caiman.add_item(
    algo="mcorr",
    input_movie_path="/home/kushal/lab/datasets/jason/61/D05/61D5_MC_mmap.tif",
    params=mcorr_params,
    item_name="dense"
)
```

```python
for i, row in df.iterrows():
    if row["outputs"] is not None: # item has already been run
        continue # skip
        
    process = row.caiman.run(wait=False)
```
